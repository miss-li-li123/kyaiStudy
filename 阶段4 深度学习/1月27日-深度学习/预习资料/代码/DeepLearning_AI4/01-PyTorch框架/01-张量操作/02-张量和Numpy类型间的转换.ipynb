```python
import torch
import numpy as np
```

```python
# 1------ 张量 转成 numpy的ndarray数组
t0 = torch.randint(low=0,high=10,size=(2,3))
t0
```

```python
# 张量.numpy()转成ndarry数据，会共享内容
# 共享内容
# arr1 = t0.numpy()

# 不共享内容。就是深拷贝
arr1 = t0.numpy().copy()
arr1
```

```python
arr1[1][1]=100
arr1
```

```python
t0
```

```python
# 2------ numpy的ndarray 转成 张量
arr2 = np.array([11,22,33])
arr2
```

```python
# torch.tensor不共享内存
t2 = torch.tensor(arr2)
t2
```

```python
t2[1]=999
t2
```

```python
arr2
```

```python
# 共享内存
t3 = torch.from_numpy(arr2)
t3
```

```python
t3[1]=999
t3
```

```python
arr2
```

```python
# 3------ 张量 转成 标量
t4 = torch.tensor(999)
t4
```

```python
result = t4.item()
print(result,type(result))
```

```python
# 注意：item()只支持张量中只有一个元素的情况。否则会报错a Tensor with 2 elements cannot be converted to Scalar
# 因此下面的代码会报错
t5 = torch.tensor([11,22])
print(t5.item())
```
