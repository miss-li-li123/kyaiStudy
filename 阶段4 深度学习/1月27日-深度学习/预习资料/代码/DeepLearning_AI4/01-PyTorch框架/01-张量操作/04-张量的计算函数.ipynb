```python
import torch
```

```python
t0 = torch.randint(low=0,high=10,size=(2,3),dtype=torch.float32)
t0
```

```python
"""
    mean(dim)：用来计算元素均值
        dim参数如果不指定，那么是对所有元素计算均值
        
        注意：用来计算均值的张量中，元素的类型必须是小数
"""
# r1 = t0.mean() # 所有元素进行计算
# r1 = t0.mean(dim=0) # 行和行之间计算均值
r1 = t0.mean(dim=1) # 列和列之间计算均值
r1
```

```python
r2 = t0.sum() # 所有元素进行计算，得到的值只有一个 # 重点掌握
# r2 = t0.sum(dim=0) # 行和行之间计算求和
# r2 = t0.sum(dim=1) # 列和列之间计算求和
r2
```
