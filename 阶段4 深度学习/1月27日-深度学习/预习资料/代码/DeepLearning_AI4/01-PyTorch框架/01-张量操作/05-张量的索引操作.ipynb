```python
import torch
```

```python
t0 = torch.randint(0,10,(4,5))
t0
```

```python
# 【掌握】1----- 简单的行、列索引 --------
# r1 = t0[0] # 获取第一行数据
# r1 = t0[:,0] # 获取第一列数据
r1 = t0[:,-1] # 获取最后一列数据
r1
```

```python
# 【掌握】2------ 范围索引：与容器中的切片操作完全一样 -------
# 需求1：前3行的前2列数据
r2 = t0[:3,:2]
r2
```

```python
# 需求2：第2行到最后的行，并且前2列的数据
r3 = t0[1:,:2]
r3
```

```python
r3 = t0[::2,:2] # 同时还支持设置步长
r3
```

```python
# 【了解】3------- 列表索引 -----
# 需求1：返回(0,1)、(1,2)两个位置的元素
#      [[行索引], [列索引]]
r4 = t0[[0,1], [1,2]]
r4
```

```python
# 需求2：获取两个元素的坐标是(2,4)、(3,3)
r5 = t0[[2,3], [4,3]]
r5
```

```python
# 获取4个元素，坐标分别是(2,4)、(2,3)、(3,4)、(3,3)
r6 = t0[[[2],[3]], [4,3]]
r6
```

```python
t0 = torch.randint(0,10,(4,5))
t0
```

```python
# 【了解】4------- 布尔索引 -------
# 需求1：第3列大于5的，对应的行数据
"""
    需求拆解：
        1- 取第3列
        2- 对取出的第3列中每个元素判断是否是大于5
        3- 取出对应的行数据
"""
# 1- 取第3列
r6 = t0[:,2]

# 2- 对取出的第3列中每个元素判断是否是大于5
r6 = t0[:,2]>5

# 3- 取出对应的行数据
r6 = t0[t0[:,2]>5]

r6
```

```python
# 需求2：第二行大于5的，对应的列数据
r7 = t0[:,t0[1]>5]
r7
```

```python
# 【了解】5------ 多维索引 --------
t1 = torch.randint(0,10,size=(3,4,5))
t1
```

```python
# 获取0轴上的数据
r1 = t1[0,:,:]
r1
```

```python
# 获取1轴上的数据
r2 = t1[:,0,:]
r2
```

```python
# 获取2轴上的数据
r3 = t1[:,:,0]
r3
```
