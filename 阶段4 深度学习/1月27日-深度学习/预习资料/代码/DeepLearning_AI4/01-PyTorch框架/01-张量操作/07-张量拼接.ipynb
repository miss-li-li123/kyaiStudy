```python
import torch
```

```python
# 1----- cat/concat ------
t0 = torch.randint(low=0,high=10,size=(2,3))
t1 = torch.randint(low=0,high=10,size=(2,3))
t1.backward()
print(t0)
print(t1)
```

```python
cat_t1 = torch.cat([t0,t1],dim=0) # 二维张量的情况下，dim=0按行进行拼接
print(cat_t1)
print(cat_t1.shape)
```

```python
cat_t1 = torch.cat([t0,t1],dim=1) # 二维张量的情况下，dim=1按行进列拼接
print(cat_t1)
print(cat_t1.shape)
```

```python
t0 = torch.randint(low=0,high=10,size=(2,3))
t1 = torch.randint(low=0,high=10,size=(4,3))

print(t0)
print(t1)
```

```python
cat_t1 = torch.cat([t0,t1],dim=0) # 二维张量的情况下，dim=0按行进行拼接
print(cat_t1)
print(cat_t1.shape)
```

```python
# 注意：cat/concat拼接的时候，拼接的指定维度/形状不用相同，其他不参与拼接的维度/形状必须相同
t0 = torch.randint(low=0,high=10,size=(2,3))
t1 = torch.randint(low=0,high=10,size=(2,4))

print(t0)
print(t1)
```

```python
cat_t1 = torch.cat([t0,t1],dim=0) # 二维张量的情况下，dim=0按行进行拼接
print(cat_t1)
print(cat_t1.shape)
```

```python
# 2----- stack --------
# 注意：两个张量的形状必须完全相同。同时拼接后的结果会进行升维
t0 = torch.randint(low=0,high=10,size=(3,4))
t1 = torch.randint(low=0,high=10,size=(3,4))

print(t0)
print(t1)
```

```python
# stack_t1 = torch.stack([t0,t1],dim=0) # 2,3,4
# stack_t1 = torch.stack([t0,t1],dim=1) # 3,2,4
stack_t1 = torch.stack([t0,t1],dim=2) # 3,4,2
print(stack_t1)
print(stack_t1.shape) 
```

```python
# 注意：两个张量的形状必须完全相同。同时拼接后的结果会进行升维
t0 = torch.randint(low=0,high=10,size=(3,5))
t1 = torch.randint(low=0,high=10,size=(3,4))
stack_t1 = torch.stack([t0,t1],dim=0)
print(stack_t1)
print(stack_t1.shape) 
```
