```python
import torch
```

```python
# 1------ 基础运算 ------
t0 = torch.randint(low=-5,high=5,size=(2,3))
t0
```

```python
# 不带下划线，那么不会对原始张量修改
t1 = t0.add(10)
print(t0)
print(t1)
```

```python
t2 = t0.sub(10)
t2
```

```python
t3 = t0.mul(10)
t3
```

```python
t4 = t0.neg()
t4
```

```python
# 带下划线，那么会对原始张量直接修改
t5 = t0.add_(10)
print(t5)
print(t0)
```

```python
# 推荐掌握直接用符号的形式
t0 = torch.randint(low=-5,high=5,size=(2,3))
# t6 = t0 + 100
# t6 = t0 * 100
t6 = t0 * -1
print(t0)
print(t6)
```

```python
t0+=10 # 等价于t0.add_(10)
t0
```

```python
# 2------ 张量点乘运算 -------
# 正确的演示：注意点乘的两个张量的形状必须完全相同
t1 = torch.tensor([[11,22,33],[44,55,66]])
t2 = torch.tensor([[1,2,3],[4,5,6]])
# t3 = t1.mul(t2)
t3 = t1*t2 # 推荐掌握
t3
```

```python
# 错误的演示：两个张量形状不同
t1 = torch.tensor([[11,22,33],[44,55,66]])
t2 = torch.tensor([[1,2],[4,5],[7,8]])
t3 = t1*t2
t3
```

```python
# 3----- 矩阵乘法运算 ------
# 矩阵乘法的前提：是前一个矩阵的列数与后一个矩阵的行数要相同
t1 = torch.tensor([[11,22,33],[44,55,66]]) # 2行3列
t2 = torch.tensor([[1],[4],[7]]) # 3行1列
t4 = t1@t2 # 推荐掌握
# t4 = t1.matmul(t2)
t4
```

```python
t1 = torch.tensor([[11,22,33],[44,55,66]])
t2 = torch.tensor([[1,2,3],[4,5,6]])
t5 = t1@t2
t5
```
